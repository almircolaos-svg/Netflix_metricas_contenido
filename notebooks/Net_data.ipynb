### Eliminar Nulos

```python
df_clean["director"].fillna("Unknown", inplace=True)
```

```python
df_clean["cast"].fillna("Unknown", inplace=True)
```

```python
df_clean["country"].fillna("Unknown", inplace=True)
```

```python
df_clean.dropna(subset=["date_added"], inplace=True)
```

```python
df_clean.dropna(subset=["rating"], inplace=True)
```

```python
df_clean.dropna(subset=["duration"], inplace=True)
```

```python
df_clean.isnull().sum()
```

### TIPOS DE DATOS

```python
df_clean["date_added"].sample(10)
```

```python
df_clean["date_added"].str.strip()
```

```python
for col in df_clean.select_dtypes(include="object").columns:
    con_espacios = df_clean[col].str.startswith(" ").sum() + \
                   df_clean[col].str.endswith(" ").sum()
    if con_espacios > 0:
        print(f" '{col}' tiene {con_espacios} valores con espacios extra")
```

```python
#limpio los espacios de date_added y lo convierto a formato fecha
df_clean["date_added"] = df_clean["date_added"].str.strip()
df_clean["date_added"] = pd.to_datetime(df_clean["date_added"], errors='coerce')
```

```python
for col in df_clean.select_dtypes(include="object").columns:
    con_espacios = df_clean[col].str.startswith(" ").sum() + \
                   df_clean[col].str.endswith(" ").sum()
    if con_espacios > 0:
        print(f" '{col}' tiene {con_espacios} valores con espacios extra")
    else:
        print(f"'{col}' no tiene valores con espacios extra")
```

```python
# Extraigo las columnas para la fecha
df_clean["year_added"] = df_clean["date_added"].dt.year
df_clean["month_added"] = df_clean["date_added"].dt.month
df_clean["month_name"] = df_clean["date_added"].dt.month_name()
```

```python
#como el mes sale alfabeticamente en ingles, creo un diccionario para mapearlo a español

meses = {
    1:"enero", 2:"febrero", 3:"marzo", 4:"abril",
    5:"mayo", 6:"junio", 7:"julio", 8:"agosto",
    9:"septiembre", 10:"octubre", 11:"noviembre", 12:"diciembre"
}

df_clean["month_name"] = df_clean["month_added"].map(meses)
```

```python
df_clean.sample(2)
```

```python
df_clean["release_year"] = df_clean["release_year"].astype(int)
```

```python
#como la duracion se separan en horas y season las voy a separar en numero y texto
df_clean["duration_value"] = df_clean["duration"].str.extract(r"(\d+)").astype(int)
df_clean["duration_unit"] = df_clean["duration"].str.extract(r"([a-zA-Z]+)")
```

```python
df_clean.head(5)
```

```python
# listed_in tiene multiples generos separados por coma
# Extraigo el genero principal (el primero)
df_clean["primary_genre"] = df_clean["listed_in"].str.split(",").str[0].str.strip()
```

```python
df_clean.sample(5)
```

```python
df_clean.drop_duplicates(inplace=True)
df_clean.duplicated().sum()
```

```python
#solo para verificar
print(f"Filas originales : {df.shape[0]}")
print(f"Filas limpias    : {df_clean.shape[0]}")
print(f"\nNulos restantes:\n{df_clean.isnull().sum()[df_clean.isnull().sum() > 0]}")
display(df_clean.sample(5))
```

### EXPORTAR

```python
import csv

df_clean = df_clean.replace("\n", " ", regex=True)

df_clean.to_csv(
    "data/netflix_clean.csv",
    index=False,
    quoting=csv.QUOTE_MINIMAL)
```

```python
# Conexion a SQL Server usando SQLAlchemy y pyodbc
server   = "DESKTOP-L8IFLL2\\ALMIRCOSQL"
database = "NetflixDB"

params = urllib.parse.quote_plus(
    f"DRIVER={{ODBC Driver 17 for SQL Server}};"
    f"SERVER={server};"
    f"DATABASE={database};"
    f"Trusted_Connection=yes;"
)

engine = create_engine(f"mssql+pyodbc:///?odbc_connect={params}")
```

```python
# CARGAR DATAFRAME A SQL
df_clean.to_sql(name = "netflix_titles", con = engine, if_exists = "replace", index = False)

print("Tabla 'netflix_titles' cargada en NetflixDB")
```
